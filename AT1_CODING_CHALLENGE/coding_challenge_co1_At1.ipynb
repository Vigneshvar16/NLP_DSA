```python
import re

TOKEN = re.compile(r"[A-Za-z]+(?:['-][A-Za-z]+)*")

def main():
    input_data = """4
The dog's tail wagged -- twice!
state-of-the-art NLP systems, circa 2026.
12345 ... ???
Don't re-use rock--roll O'Neill's x"""

    data = input_data.splitlines()
    n = int(data[0])
    out = []

    for i in range(1, n + 1):
        line = data[i]
        toks = TOKEN.findall(line)
        out.append(" ".join([str(len(toks))] + toks))

    print("\n".join(out))

main()
```

```python

VOWELS = set("aeiou")

# Lexicon: irregular word forms
LEX = {
    ("goose", "N+PL"): "geese",
    ("go", "V+PAST"): "went",
    ("good", "ADJ+COMP"): "better",
}


def is_cons(ch):
    return ch not in VOWELS


def vowel_groups(word):
    groups = 0
    prev = False

    for ch in word:
        vowel = ch in VOWELS
        if vowel and not prev:
            groups += 1
        prev = vowel

    return groups


def doublable(word):
    return (
        len(word) >= 3
        and is_cons(word[-3])
        and not is_cons(word[-2])
        and is_cons(word[-1])
        and word[-1] not in "wxy"
        and vowel_groups(word) == 1
    )


def cons_then(word, ch):
    return len(word) >= 2 and word[-1] == ch and is_cons(word[-2])


def inflect(stem, feat):
    if feat in ("N+PL", "V+3SG"):
        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"
        if cons_then(stem, "o"):
            return stem + "es"
        if cons_then(stem, "y"):
            return stem[:-1] + "ies"
        return stem + "s"

    if feat == "V+PROG":
        if len(stem) >= 3 and cons_then(stem, "e"):
            return stem[:-1] + "ing"
        if doublable(stem):
            return stem + stem[-1] + "ing"
        return stem + "ing"

    suffixes = {
        "V+PAST": ("d", "ed"),
        "ADJ+COMP": ("r", "er"),
        "ADJ+SUP": ("st", "est"),
    }

    if feat in suffixes:
        short, full = suffixes[feat]

        if stem.endswith("e"):
            return stem + short
        if cons_then(stem, "y"):
            return stem[:-1] + "i" + full
        if doublable(stem):
            return stem + stem[-1] + full

        return stem + full

    return stem


def main():
    # Input included directly in the program
    data = """
    3
    goose N+PL geese
    go V+PAST went
    good ADJ+COMP better
    8
    cat N+PL
    church N+PL
    baby N+PL
    stop V+PROG
    make V+PROG
    carry V+PAST
    goose N+PL
    go V+PAST
    """

    tokens = data.split()
    p = 0

    L = int(tokens[p])
    p += 1

    lex = {}

    for _ in range(L):
        stem, feat, form = tokens[p:p + 3]
        lex[(stem, feat)] = form
        p += 3

    Q = int(tokens[p])
    p += 1

    output = []

    for _ in range(Q):
        stem, feat = tokens[p:p + 2]
        p += 2

        result = lex.get((stem, feat))
        if result is None:
            result = inflect(stem, feat)

        output.append(result)

    print("\n".join(output))


if __name__ == "__main__":
    main()
```

```python

VOWELS = set("aeiou")


def is_cons(ch):
    return ch not in VOWELS


def vowel_groups(w):
    g, prev = 0, False
    for ch in w:
        v = ch in VOWELS
        if v and not prev:
            g += 1
        prev = v
    return g


def doublable(w):
    return (
        len(w) >= 3
        and is_cons(w[-3])
        and not is_cons(w[-2])
        and is_cons(w[-1])
        and w[-1] not in "wxy"
        and vowel_groups(w) == 1
    )


def cons_then(w, ch):
    return len(w) >= 2 and w[-1] == ch and is_cons(w[-2])


def inflect(stem, feat):
    if feat in ("PL", "3SG"):
        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"
        if cons_then(stem, "o"):
            return stem + "es"
        if cons_then(stem, "y"):
            return stem[:-1] + "ies"
        return stem + "s"

    if feat == "PROG":
        if len(stem) >= 3 and cons_then(stem, "e"):
            return stem[:-1] + "ing"
        if doublable(stem):
            return stem + stem[-1] + "ing"
        return stem + "ing"

    short, full = {
        "PAST": ("d", "ed"),
        "COMP": ("r", "er"),
        "SUP": ("st", "est"),
    }[feat]

    if stem.endswith("e"):
        return stem + short
    if cons_then(stem, "y"):
        return stem[:-1] + "i" + full
    if doublable(stem):
        return stem + stem[-1] + full

    return stem + full


FEATURES = {
    "N": ["PL"],
    "V": ["3SG", "PROG", "PAST"],
    "ADJ": ["COMP", "SUP"],
}


def main():
    data = """
    5
    cat N
    watch N
    watch V
    study V
    big ADJ
    8
    cats
    watches
    watching
    studied
    bigger
    cat
    biggest
    elephants
    """

    tok = data.split()
    p = 0

    S = int(tok[p])
    p += 1

    entries = set()

    for _ in range(S):
        entries.add((tok[p], tok[p + 1]))
        p += 2

    table = {}

    for stem, pos in entries:
        table.setdefault(stem, set()).add(stem + "+" + pos)

        for feat in FEATURES.get(pos, []):
            word = inflect(stem, feat)
            table.setdefault(word, set()).add(
                stem + "+" + pos + "+" + feat
            )

    Q = int(tok[p])
    p += 1

    out = []

    for _ in range(Q):
        word = tok[p]
        p += 1

        result = table.get(word)
        out.append(", ".join(sorted(result)) if result else "UNKNOWN")

    print("\n".join(out))


if __name__ == "__main__":
    main()
```

```python

def cons(w, i):
    ch = w[i]
    if ch in "aeiou":
        return False
    if ch == "y":
        return True if i == 0 else not cons(w, i - 1)
    return True


def measure(w):
    m, prev_vowel = 0, False
    for i in range(len(w)):
        c = cons(w, i)
        if c and prev_vowel:
            m += 1
        prev_vowel = not c
    return m


def has_vowel(w):
    return any(not cons(w, i) for i in range(len(w)))


def ends_double_cons(w):
    return len(w) >= 2 and w[-1] == w[-2] and cons(w, len(w) - 1)


def ends_cvc(w):
    n = len(w)
    return (
        n >= 3
        and cons(w, n - 3)
        and not cons(w, n - 2)
        and cons(w, n - 1)
        and w[-1] not in "wxy"
    )


def step1a(w):
    if w.endswith("sses"):
        return w[:-2]
    if w.endswith("ies"):
        return w[:-2]
    if w.endswith("ss"):
        return w
    if w.endswith("s"):
        return w[:-1]
    return w


def cleanup(s):
    if s.endswith(("at", "bl", "iz")):
        return s + "e"
    if ends_double_cons(s) and s[-1] not in "lsz":
        return s[:-1]
    if measure(s) == 1 and ends_cvc(s):
        return s + "e"
    return s


def step1b(w):
    if w.endswith("eed"):
        stem = w[:-3]
        return stem + "ee" if measure(stem) > 0 else w

    for suf in ("ed", "ing"):
        if w.endswith(suf):
            stem = w[:-len(suf)]
            return cleanup(stem) if has_vowel(stem) else w

    return w


def step1c(w):
    if w.endswith("y") and has_vowel(w[:-1]):
        return w[:-1] + "i"
    return w


def main():
    data = """
    8
    caresses
    ponies
    cats
    feed
    agreed
    motoring
    hopping
    happy
    """

    tok = data.split()
    q = int(tok[0])

    out = [
        step1c(step1b(step1a(w)))
        for w in tok[1:1 + q]
    ]

    print("\n".join(out))


if __name__ == "__main__":
    main()
```
