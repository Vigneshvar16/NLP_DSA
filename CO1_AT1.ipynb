```python
import re

# ---------- Sample data ----------
resumes = [
"""Name: Rahul Kumar
Email: rahul.kumar@gmail.com
Mobile: +91 98765 43210
Skills: Python, SQL, NLP
Experience: 3 years""",

"""Name: Priya Sharma
Email: priya_sharma@outlook.com
Mobile: 91234 56789
Skills: JAVA, sql
Experience: 1 year""",

"""Name: Ankit Verma
Email: ankit.verma@company.co.in
Mobile: 98765-43210
Skills: PYTHON, Machine Learning, nlp
Experience: 5 years""",

"""Name: Sneha Iyer
Email: sneha.iyer@mail.com
Mobile: +91-9012345678
Skills: python, Java
Experience: 1 year""",
]

# ---------- Regex patterns ----------
NAME = r"Name:\s*([A-Za-z ]+)"
EMAIL = r"[\w.+-]+@[\w-]+\.[\w.]+"
MOBILE = r"(?:\+?91[\s-]?)?[6-9]\d{4}[\s-]?\d{5}"
SKILL = r"\b(Python|Java|SQL|Machine Learning|NLP)\b"
EXPERIENCE = r"(\d+)\s+years?"

PROPER = {"python": "Python", "java": "Java", "sql": "SQL",
          "machine learning": "Machine Learning", "nlp": "NLP"}


def extract(text):
    name = re.search(NAME, text)
    email = re.search(EMAIL, text)
    mobile = re.search(MOBILE, text)
    exp = re.search(EXPERIENCE, text)
    skills = []
    for s in re.findall(SKILL, text, re.IGNORECASE):
        s = PROPER[s.lower()]
        if s not in skills:
            skills.append(s)
    return {
        "name": name.group(1).strip() if name else "Not found",
        "email": email.group() if email else "Not found",
        "mobile": mobile.group() if mobile else "Not found",
        "skills": skills,
        "experience": int(exp.group(1)) if exp else 0,
    }


print("RESUME INFORMATION EXTRACTION\n")
eligible = []
for resume in resumes:
    c = extract(resume)
    print("Candidate:", c["name"])
    print("Email:", c["email"])
    print("Mobile:", c["mobile"])
    print("Skills:", ", ".join(c["skills"]))
    unit = "year" if c["experience"] == 1 else "years"
    print("Experience:", c["experience"], unit)
    print()
    if c["experience"] >= 2 and "Python" in c["skills"]:
        eligible.append(c["name"])

print("ELIGIBLE CANDIDATES:")
for i, name in enumerate(eligible, 1):
    print(f"{i}. {name}")

print("\nExtraction completed successfully.")
```

```python
import re

# ---------- Sample data ----------
products = [
    "Apple iPhone 15",
    "Samsung Galaxy S24",
    "Dell Laptop",
    "HP Laptop",
    "Apple MacBook Air",
    "Wireless Mouse",
    "Gaming Keyboard",
    "Bluetooth Speaker",
    "Sony Headphone",
    "USB Cable",
    "Realme Smartphone",
    "Laptop Bag",
]


# ---------- Search patterns ----------
def exact(k):
    return r"\b" + re.escape(k) + r"\b"    # whole word


def prefix(k):
    return r"\b" + re.escape(k)            # word starts with keyword


def suffix(k):
    return re.escape(k) + r"\b"            # word ends with keyword


def partial(k):
    return re.escape(k)                    # keyword anywhere


def search(pattern, flags=re.IGNORECASE):
    return [p for p in products if re.search(pattern, p, flags)]


def show(title, keyword, pattern):
    found = search(pattern)
    print(f'{title}: "{keyword}"')
    print("Matches:", len(found))
    print("  ->", ", ".join(found) if found else "None")
    print()


print("PRODUCT SEARCH SYSTEM\n")
show("Exact search", "laptop", exact("laptop"))
show("Prefix search", "lap", prefix("lap"))
show("Suffix search", "phone", suffix("phone"))
show("Partial search", "book", partial("book"))

print('Case-insensitive search: "laptop" / "Laptop" / "LAPTOP"')
counts = [len(search(exact(w))) for w in ("laptop", "Laptop", "LAPTOP")]
print("Matches:", counts[0], "/", counts[1], "/", counts[2])
print("  -> Same result for all cases:", len(set(counts)) == 1)
print('  -> Without re.IGNORECASE, "LAPTOP" gives:',
      len(search(exact("LAPTOP"), 0)))

print("\nSearch completed successfully.")
```

```python
import re

# ---------- Sample data ----------
students = [
    {   # valid record
        "name": "Arun Kumar",
        "register": "24CSE101",
        "email": "arun.kumar@college.edu",
        "course": "CS101",
        "semester": "SEM1",
        "mobile": "9876543210",
    },
    {   # invalid record
        "name": "Meena Rao",
        "register": "24CS1",
        "email": "meena@gmail.com",
        "course": "CS205",
        "semester": "SEM9",
        "mobile": "9123456780",
    },
]


# ---------- Validation functions ----------
def validate_register_number(value):
    return bool(re.fullmatch(r"\d{2}[A-Z]{2,3}\d{3}", value))   # 24CSE101


def validate_email(value):
    return bool(re.fullmatch(r"[A-Za-z0-9._]+@college\.edu", value))


def validate_course_code(value):
    return bool(re.fullmatch(r"[A-Z]{2,3}\d{3}", value))        # CS101


def validate_semester(value):
    return bool(re.fullmatch(r"SEM[1-8]", value))               # SEM1-SEM8


def validate_mobile(value):
    return bool(re.fullmatch(r"[6-9]\d{9}", value))             # 10 digits


checks = [
    ("Register Number", "register", validate_register_number),
    ("Email", "email", validate_email),
    ("Course Code", "course", validate_course_code),
    ("Semester", "semester", validate_semester),
    ("Mobile Number", "mobile", validate_mobile),
]

print("UNIVERSITY REGISTRATION VALIDATION")
for s in students:
    print("\nStudent:", s["name"])
    all_valid = True
    for label, key, func in checks:
        ok = func(s[key])
        all_valid = all_valid and ok
        print(f"{label}: {'VALID' if ok else 'INVALID'}")
    status = "SUCCESSFUL" if all_valid else "UNSUCCESSFUL"
    print("FINAL REGISTRATION STATUS:", status)
```
