```python
import re
from collections import Counter

paragraph = input("Enter a paragraph: ")

# Split into sentences
sentences = re.split(r'[.!?]+', paragraph)
sentences = [s.strip() for s in sentences if s.strip()]

# Split into words
words = re.findall(r'\b\w+\b', paragraph.lower())

# Count words
word_count = len(words)

# Find most frequent words
frequency = Counter(words)

print("\nSentences:")
for sentence in sentences:
    print(sentence)

print("\nWords:")
print(words)

print("\nTotal number of words:", word_count)

print("\nMost frequent words:")
for word, count in frequency.most_common():
    print(word, ":", count)
```

```python
def check_string(s):
    state = 0

    for ch in s:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0

        elif state == 1:
            if ch == 'a':
                state = 1
            else:
                state = 2

        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0

    return state == 2


s = input("Enter a string: ")

if check_string(s):
    print("Accepted")
else:
    print("Rejected")
```

```python
import re

text = input("Enter a string: ")

patterns = {
    "*": r"a*",
    "+": r"a+",
    "?": r"a?",
    "[]": r"[abc]",
    "^": r"^Call",
    "$": r"world$"
}

for operator, pattern in patterns.items():
    if re.search(pattern, text):
        print(operator, ": Yes")
    else:
        print(operator, ": No")
```

```python
import re

text = "Call me at 9876543210 or email john123@gmail.com on 12-10-2026."

# Find first digit
digit = re.search(r'\d', text)
print("First digit:", digit.group())

# Find mobile numbers
mobile = re.findall(r'\b\d{10}\b', text)
print("Mobile numbers:", mobile)

# Find email IDs
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', text)
print("Email IDs:", emails)

# Find capitalized words
capitalized = re.findall(r'\b[A-Z][a-z]*\b', text)
print("Capitalized words:", capitalized)

# Check whether text starts with "Call"
if re.match(r'^Call', text):
    print("Starts with Call: Yes")
else:
    print("Starts with Call: No")

# Change date format
new_text = re.sub(r'(\d{2})-(\d{2})-(\d{4})',
                  r'\3-\2-\1', text)
print("Changed date:", new_text)

# Hide phone numbers
hidden = re.sub(r'\b\d{10}\b', 'XXXXXXXXXX', text)
print("Hidden phone number:", hidden)
```
